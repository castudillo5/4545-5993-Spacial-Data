## Q6. Plotting
Create a line plot of $y = x^2$ for $x$ values from 0 to 10.

```python
import matplotlib.pyplot as plt
import numpy as np

# Create x values from 0 to 10
x = np.linspace(0, 10, 100)

# Calculate y = x²
y = x ** 2

# Create the plot
plt.plot(x, y, color='blue', linewidth=2)
plt.title("y = x²")
plt.xlabel("x")
plt.ylabel("y = x²")
plt.grid(True)
plt.show()
```
